```python
import torch
from transformers import GPT2LMHeadModel,AutoTokenizer

torch.manual_seed(12046)

tokenizer = AutoTokenizer.from_pretrained("gpt2")
model = GPT2LMHeadModel.from_pretrained("gpt2")
```

```python
# 使用分词器对文本进行分词
question='What is the captial of China?'
ids=tokenizer(question,return_tensors="pt")
ids
```

```python
# 由于GPT-2的模型效果较差，通过增大num_beams和no_repeat_ngram_size来优化生成的文本。
res=model.generate(**ids,max_length=100,early_stopping=True,num_beams=3,no_repeat_ngram_size=2)
print(res[0])
print(tokenizer.decode(res[0],skip_special_tokens=True))
```
